### Processing image files from the extracted dataset

Now, let's adapt the face detection code to process the image files found in the `dataset/faces` directory. We'll iterate through each image, load it, and apply face detection.

```python
import os
import cv2
import numpy as np

# Define the path for the cascade classifier
HAARCASCADE_PATH = 'data/haarcascade_frontalface_default.xml'

# Check if the 'data' directory exists, if not, create it
if not os.path.exists('data'):
    os.makedirs('data')

# Check if the cascade classifier file exists, if not, download it
if not os.path.exists(HAARCASCADE_PATH):
    print(f"Downloading {HAARCASCADE_PATH}...")
    # Using wget to download the file from OpenCV's GitHub
    !wget -P data/ https://raw.githubusercontent.com/opencv/opencv/4.x/data/haarcascades/haarcascade_frontalface_default.xml
    print("Download complete.")

# Ensure facedetect is loaded
facedetect = cv2.CascadeClassifier(HAARCASCADE_PATH)
if facedetect.empty():
    print(f"Error: Could not load cascade classifier. Check path: {HAARCASCADE_PATH}")
    # Exit or raise an error if the classifier still can't be loaded
    # For now, we'll just print an error and continue, but subsequent calls to detectMultiScale will fail.

faces_data = []
i = 0

# Path to the directory containing face images
faces_dir = 'dataset/faces'

# Iterate through each person's folder
for person_name in os.listdir(faces_dir):
    person_dir = os.path.join(faces_dir, person_name)
    if os.path.isdir(person_dir):
        print(f"Processing images for: {person_name}")
        # Iterate through each image in the person's folder
        for image_name in os.listdir(person_dir):
            image_path = os.path.join(person_dir, image_name)

            # Read the image
            frame = cv2.imread(image_path)

            if frame is None:
                print(f"Warning: Could not read image {image_path}. Skipping.")
                continue

            gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            # Check if facedetect is loaded before calling detectMultiScale
            if not facedetect.empty():
                faces = facedetect.detectMultiScale(gray, 1.3, 5)
            else:
                print("Cascade classifier not loaded, skipping face detection for this image.")
                faces = [] # Assign an empty list to avoid further errors

            for (x, y, w, h) in faces:
                crop_img = frame[y:y+h, x:x+w, :]
                resized_img = cv2.resize(crop_img, (50, 50))
                if len(faces_data) <= 100 and i % 10 == 0:
                    faces_data.append(resized_img)
                i += 1

                # Optional: Display detection on the image (remove for large datasets)
                # cv2.putText(frame, str(len(faces_data)), (50, 50), cv2.FONT_HERSHEY_COMPLEX, 1, (50, 50, 255), 1)
                # cv2.rectangle(frame, (x, y), (x+w, y+h), (50, 50, 255), 1)

            # Optional: Display the image with detections (remove for large datasets)
            # cv2.imshow(f"Face Detection - {person_name}", frame)
            # k = cv2.waitKey(1) # Wait for 1ms
            # if k == ord('q'):
            #     break

# Print how many faces were collected
print(f"Collected {len(faces_data)} face samples.")

# Optional: Clean up windows if they were used
# cv2.destroyAllWindows()
```

### Displaying a few collected face samples

Let's visualize some of the face samples that were collected from the images to ensure the process worked as expected.

```python
import matplotlib.pyplot as plt

if len(faces_data) == 0:
    print("No face samples were collected. Please check the dataset and face detection process.")
else:
    # Display the first 5 collected faces, or fewer if not enough samples
    num_faces_to_display = min(5, len(faces_data))
    fig, axes = plt.subplots(1, num_faces_to_display, figsize=(15, 3))

    if num_faces_to_display == 1:
        axes = [axes] # Make it iterable if only one subplot

    for idx in range(num_faces_to_display):
        # OpenCV reads images as BGR, but matplotlib expects RGB
        face_rgb = cv2.cvtColor(faces_data[idx], cv2.COLOR_BGR2RGB)
        axes[idx].imshow(face_rgb)
        axes[idx].set_title(f"Face {idx+1}")
        axes[idx].axis('off')

    plt.tight_layout()
    plt.show()
```
